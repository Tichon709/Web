```python
from sympy import *
x = symbols('x')
y = Function('y')

# а)
f = (3*x + 8)**4 - cos(exp(x) - 8) + 16
print('a', diff(f, x))
# б)
eq = Eq(diff(y(x), x), 2*y(x)/x)
print("Общее решение:", dsolve(eq, y(x)))
print("Решение с начальным условием:", dsolve(eq, y(x), ics={y(2): 1}))

```

```python
x = symbols('x', positive=True)
y = Function('y')
C1 = symbols('C1')
# a
f = -1 + exp(-x)*cos(x + 2)
print("a", diff(f, x))
# b
eq = Eq(diff(y(x), x), 4*x*log(x))
solution = dsolve(eq, y(x))
C = solve(Eq(limit(solution.rhs, x, 0, dir='+'), 1), C1)[0]
print("b", solution.subs(C1, C))
```

```python
from sympy import *
x = symbols('x')
y = Function('y')
C1 = symbols('C1')

# a
f = 7*x**3*cos(2*x)
print("a", diff(f, x))

# b
eq = Eq((x + 1)*diff(y(x), x), -x*y(x))
solution = dsolve(eq, y(x))
C = solve(Eq(solution.rhs.subs(x, 0), 1), C1)[0]
print("b", solution.subs(C1, C))
```
